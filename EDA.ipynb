```python
# -------------------------------
# Explore & clean columns
# -------------------------------
data = pd.read_csv('data/raw/k-pro-db.tsv', sep = '\t')

# 1. Protein
data['PROTEIN'] = data['PROTEIN'].astype('category')

# 2. Source
data['SOURCE'] = data['SOURCE'].astype('category')

# Null values is removed as there is only 2 and they represent protein missing significant amounts of data which cannot be interpolated
data = data[~data['SOURCE'].isnull()]               

# 3. Length
data['LENGTH'] = data['LENGTH'].astype('int32')

# 4. Uniprot
# Input UNIPROT ID from other entries of the same protein when missing
data.loc[data['UniProt'].isnull() & data['PROTEIN'].str.contains('PSD1'), 'UniProt'] = 'Q51918' 
data['UniProt'] = data['UniProt'].astype('category')

# 5. PDB wild
data['PDB_wild'] = data['PDB_wild'].astype('category')

# 6. Mutated chain
data['MUTATED_CHAIN'] = data['MUTATED_CHAIN'].astype('category')

# 7. Class (Nan values remain to be handled)
data['CLASS'] = data['CLASS'].replace({'Alpha Beta/Mainly Alpha/Mainly Beta': 'Alpha Beta'})
data['CLASS'] = data['CLASS'].astype('category')

# 8. PFAM
#print(data.loc[data['PROTEIN'].str.contains('Rd-apocyt b562'), 'PFAM'])
#print(data.loc[data['PFAM'].isnull()])
#print(data['PFAM'].value_counts(dropna = False))

print(data['UniProt'].value_counts())
# 10. EC number (Not needed)
#data.drop(columns = ['EC_NUMBER', 'ION_NAME', 'ION_CONC', 'STATE', 'AUTHOR', 'REFERENCE', 'Year', 'PMID'])

# 9. 
print(data.loc[:,['PROTEIN', 'SOURCE', 'LENGTH', 'UniProt', 'PDB_wild', 'MUTATED_CHAIN', 'CLASS']].isnull().sum())
```
